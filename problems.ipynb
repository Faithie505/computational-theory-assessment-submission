## Problem 1: Representing SHA-256 Data

SHA-256 is a Hashing Algorithm commonly used in cryptography. After it is fed data, it converts that data into a 256 bit encoded hash value.  
To use the SHA-256 algorithm in Python, the main things to be considered are:
- The input: What we want to send into the algorithm
- The encoding 

References:
https://www.qcecuring.com/blog/sha-256


```python
import hashlib

text = "Hello, world. This is a test!"
digest = hashlib.sha256(text.encode("utf-8")).hexdigest();
print(digest)
```
