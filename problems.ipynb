## Problem 1: Representing SHA-256 Data

SHA-256 is a Hashing Algorithm commonly used in cryptography. It is a part of Pythons hashlib library ([official documentation](https://docs.python.org/3/library/hashlib.html)). The algorithm takes in an input, which then converts its data into a 256 bit encoded hash value. As a hashing algorithm, the converted data cannot be reverse engineered or decrypted back into its original message.

To use the SHA-256 algorithm in Python, the main things to be considered are:
- The input: What data we want to send into the algorithm
- The encoding
