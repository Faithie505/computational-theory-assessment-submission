## Problem 1: Representing SHA-256 Data

SHA-256 is a Hashing Algorithm commonly used in cryptography. It is a part of Pythons hashlib library ([official documentation](https://docs.python.org/3/library/hashlib.html)). 

The algorithm takes in an input, and then converts its data into a 256 bit encoded hash value. As SHA-256 uses hashing rather than encryption, the converted data cannot be reverse engineered or decrypted back into its original message.

To use the SHA-256 algorithm in Python, the main things to be considered are:
- The input: What data we want to send into the algorithm
- The output: Whether we want the returned value to be a string of hexidecimals, or a bytes object

References:
https://www.qcecuring.com/blog/sha-256


```python
import hashlib

#initialising the variable we want to hash
text = "Hello, world"
```

```python
def returnHex():
    """Prints out a string of hexideciamal values corresponding to the message"""
    hexMessage = hashlib.sha256(text.encode()).hexdigest()
    print("The hexadecimal message is: ", hexMessage)
```

```python
def returnBytes():
    """Prints out a bytes object corresponding to the message"""
    bytesMessage = hashlib.sha256(text.encode()).digest()
    print("The bytes object message is: ", bytesMessage) 
```

```python
#calling both methods
returnHex()
returnBytes()
```
